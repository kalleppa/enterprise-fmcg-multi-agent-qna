```python
manager = None

if LIVE_RUN:
    from src.memory.conversation_manager import ConversationManager
    manager = ConversationManager()


def display_payload(payload: dict[str, Any]) -> None:
    response = payload.get("response", {})
    display(Markdown(
        f"**Question:** {payload.get('question', '')}  \n"
        f"**Execution mode:** `{payload.get('execution_mode', 'live_agent')}`  \n"
        f"**Route:** `{response.get('route', 'n/a')}`  \n"
        f"**Status:** `{response.get('status', 'n/a')}`  \n"
        f"**Context applied:** `{payload.get('context_applied', False)}`"
    ))
    display(Markdown(response.get("answer", "")))
    display(JSON(payload, expanded=False))


def run_case(case_id: str, question: str, session_id: str | None = None) -> dict[str, Any]:
    if LIVE_RUN:
        assert manager is not None
        result = manager.answer(question=question, session_id=session_id)
        payload = result.to_dict()
        payload["execution_mode"] = "live_agent"
    else:
        payload = SAVED_RESPONSES[case_id]

    display_payload(payload)
    return payload
```

## 11. Multi-turn context preservation

```python
if LIVE_RUN:
    assert manager is not None
    first = manager.answer("Show net revenue by region for Q2 2025")
    second = manager.answer("Why?", session_id=first.session_id)
    third = manager.answer("Was inventory a factor there?", session_id=first.session_id)

    multi_turn_result = {
        "execution_mode": "live_agent",
        "session_id": first.session_id,
        "turns": [first.to_dict(), second.to_dict(), third.to_dict()],
    }
else:
    multi_turn_result = MULTI_TURN_SNAPSHOT

for index, turn in enumerate(multi_turn_result["turns"], start=1):
    route = turn.get("route") or turn.get("response", {}).get("route")
    status = turn.get("status") or turn.get("response", {}).get("status")
    display(Markdown(
        f"### Turn {index}  \n"
        f"**User:** {turn.get('question') or turn.get('original_question')}  \n"
        f"**Resolved:** {turn.get('resolved_question')}  \n"
        f"**Context applied:** `{turn.get('context_applied')}`  \n"
        f"**Route / status:** `{route}` / `{status}`"
    ))

display(JSON(multi_turn_result, expanded=False))
```
